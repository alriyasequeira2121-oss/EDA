```python
import pandas as pd
```

```python
df=pd.read_csv("stock_analysis.csv")
```

```python
df
```

```python
df.info()
```

```python
df["Date"]=pd.to_datetime(df["Date"])
df.dtypes
```

```python
#sort values by date
df=df.sort_values("Date")
```

```python
df
```

```python
#moving average
df["MA_20"]=df["Close"].rolling(20).mean()
df["MA_50"]=df["Close"].rolling(50).mean()
df
```

```python
df["Daily_Return"]=(df["Close"]-df["Open"]/df["Open"])*100
df
```

```python
#analys stock return distribution

mean_return=df["Daily_Return"].mean()
print("mean_return",mean_return)

variance_return=df["Daily_Return"].var()  
print("variance_return",variance_return)

standard_return=df["Daily_Return"].std()
print("standard_return",standard_return)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12,6))
plt.plot(df["Date"],df["Open"],label="Open")
plt.plot(df["Date"],df["Close"],label="Close")
plt.plot(df["Date"],df["High"],label="High")
plt.plot(df["Date"],df["Low"],label="Low")
plt.title("OCHL prices over time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df["Date"],df["Volume"],label="Volume")
plt.title("trading volume prices over time")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,6))
plt.plot(df["Date"],df["MA_20"],label="MA_20")
plt.plot(df["Date"],df["Close"],label="Close")
plt.plot(df["Date"],df["MA_50"],label="MA_50")
plt.title("moving average prices over time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df["Daily_Return"],bins=30)
plt.title("Distribution of Daily returns")
plt.xlabel("Daily_returns")
plt.ylabel("Frequercy")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df["Daily_Return"],bins=30,kde=True,color="BLUE")
plt.title("Distribution of Daily returns")
plt.xlabel("Daily_returns")
plt.ylabel("Frequency")
plt.xticks(rotation=35)
plt.show()
```

```python
Daily_Return=df["Daily_Return"]
avg_daily_return=Daily_Return.mean()
variance=Daily_Return.var()
std_deviation=Daily_Return.std()
max_return=Daily_Return.max()
min_return=Daily_Return.min()

print("FINANCIAL SUMMARY")
print("-"*30)
print("AVERAGE DAILY RETURNS:",avg_daily_return)
print("VARIANCE:",variance)
print("STANDARD DEVIATION:",std_deviation)
print("MAXIMUM DAILY RETURN:",max_return)
print("MINIMUM DAILY RETURN:",min_return)
```
