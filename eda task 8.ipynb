```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("StudentsPerformance.csv")
```

```python
#grouped barchart
parental_avg=df.groupby("parental level of education")[
    ["math score","reading score","writing score"]
    ].mean()

print("\Average scores by parental education:")
print(parental_avg)

parental_avg.plot(
    kind="bar",
    figsize=(12,6)
)
plt.title("Average Test Scores by Parental Education Level")
plt.xlabel("Parental Education Level")
plt.ylabel("Average Score")
plt.xticks(rotation=30)
plt.legend(title="Subject")
plt.show()
```

```python
lunch_avg=df.groupby("lunch")[
    ["math score","reading score","writing score"]
    ].mean()

print("\Average scores by lunch:")
print(lunch_avg)

lunch_avg.plot(
    kind="bar",
    figsize=(12,6)
)
plt.title("Average Test Scores by lunch Level")
plt.xlabel("lunch")
plt.ylabel("Average Score")
plt.xticks(rotation=30)
plt.legend(title="Subject")
plt.show()
```

```python
#boxplot
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="test preparation course",
    y="reading score"
)

plt.title("Reading Score by Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Reading Score")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="test preparation course",
    y="math score"
)

plt.title("math score by Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("math score")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="test preparation course",
    y="math score"
)

plt.title("math score by Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("math score")
plt.show()
```

```python
#heatmap
numeric_cols=["math score","reading score","writing score"]
alriya=df[numeric_cols].corr()
alriya
```

```python
sns.heatmap(alriya,
           annot=True,
           cmap="coolwarm",
           cbar=False,
           square=True,
           linewidth=1)
plt.title("total score")
plt.show()
```

```python
sns.scatterplot(x="math score",y="reading score",data=df)
plt.xlabel("math score")
plt.ylabel("reading score")
plt.title("math score v/s reading score")
plt.grid(True)
plt.show()  
```

```python
sns.scatterplot(x="math score",y="writing score",data=df)
plt.xlabel("math score")
plt.ylabel("writing score")
plt.title("math score v/s writing score")
plt.grid(True)
plt.show()  
```

```python
sns.scatterplot(x="writing score",y="reading score",data=df)
plt.xlabel("writing score")
plt.ylabel("reading score")
plt.title("writing score v/s reading score")
plt.grid(True)
plt.show()  
```
