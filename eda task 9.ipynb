```python
df["date"]=pd.to_datetime(df["date"],errors='coerce')
df.dtypes
```

```python
df.isnull().sum()
```

```python
print("Zero values in bedrooms:",(df["bedrooms"]==0).sum())
print("Zero values in bathrooms:",(df["bathrooms"]==0).sum())
```

```python
import numpy as np
```

```python
df["bedrooms"]=df["bedrooms"].replace(0,np.nan)
df["bathrooms"]=df["bathrooms"].replace(0,np.nan)
```

```python
df.isnull().sum()
```

```python
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].mean())
```

```python
df['bathrooms'] = df['bathrooms'].fillna(df['bedrooms'].median())
```

```python
df.isnull().sum()
```

```python
#exract year
df["sale_year"]=df["date"].dt.year

#calculate age of house at time of sale
df["house_age"]=df["sale_year"]-df["yr_built"]

df["year_since_renovation"]=np.where(
    df["yr_renovated"]>0,
    df["sale_year"]-df["yr_renovated"],
    np.nan
)
df[["date","sale_year","yr_built","yr_renovated",
    "house_age","year_since_renovation"]].head()
```

```python
df["price_per_sqft_living"]=df["price"]/df["sqft_living"]
df"[price_per_sqft_lot"]=df["price"]/df["sqft_lot"]
```

```python
df[["price","sqft_living","sqft_lot",
    "price_per_sqft_living","price_per_sqft_lot"]].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(x=df["price"])
plt.title("box plot of house price")
plt.xlabel("price")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(x=df["price_per_sqft_living"])
plt.title("box plot of price_per_sqft_living")
plt.xlabel("price per sqft living")
plt.show()
```

```python
Q1=df["price"].quantile(0.25)
Q3=df["price"].quantile(0.75)

IQR=Q3-Q1

lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("lower_bound:",lower_bound)
print("upper_bound:",upper_bound)
```

```python
print_outliers=df[
    (df["price"]<lower_bound)|
    (df["price"]>upper_bound)]

print("Number of price outlier:",len(print_outliers))

print_outliers[["id","price","bedrooms",
                "bathrooms","sqft_living"]].head(20)
    
```
