```python
df["Discharge Date"]=pd.to_datetime(df["Discharge Date"],errors='coerce')
df.dtypes
```

```python
df["Date of Admission"]=pd.to_datetime(df["Date of Admission"],errors='coerce')
df.dtypes
```

```python
df["Admission Type"].value_counts()
```

```python
df["Medical Condition"].value_counts()
```

```python
def categories_level(admission):
    if admission=='Emergency':
        return 'immediate care'
    elif admission=='Urgent':
        return 'hign priority'
    else:
        return 'planned care'
df["priority_level"]=df["Admission Type"].apply(categories_level)
```

```python
df
```

```python
df["days in hospital"]=(df["Discharge Date"]-df["Date of Admission"]).dt.days
```
