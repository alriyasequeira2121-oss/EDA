```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("kc_house_data.csv")
```

```python
df
```

```python
plt.figure(figsize=(12,6))

sns.regplot(
    data=df,
    x="sqft_living",
    y="price",
)

plt.title("price vs sqft_living	")
plt.xlabel("price per sqft living")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(
    data=df,
    x="grade",
    y="price"
)

plt.title("grade vs price")
plt.xlabel("grade")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(
    data=df,
    x="waterfront",
    y="price"
)

plt.title("waterfront vs price")
plt.xlabel("waterfront")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.scatterplot(
    data=df,
    x="long",
    y="lat",
)

plt.title("latitude and longitude ")
plt.xlabel("longitude")
plt.ylabel("latitude")
plt.legend=False
plt.palette="blue"
plt.show()
```
