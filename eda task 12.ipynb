```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("Automobile_data.csv")
df
```

```python
df=df.replace('?',np.nan)
print(df.head())
```

```python
# Numerical columns
num_cols = ['normalized-losses', 'bore', 'stroke', 
            'horsepower', 'peak-rpm', 'price']

# Categorical columns
cat_cols = ['num-of-doors']

# Fill numerical null values with median
for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].fillna(df[col].median())

# Fill categorical null values with mode
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
numeric_columns = [
    "horsepower",
    "engine-size",
    "city-mpg",
    "price",
    "curb-weight",
    "width",
    "length",
    "wheel-base",
    "bore",
    "stroke",
    "compression-ratio",
    "peak-rpm",
    "highway-mpg",
    "height",
    "symboling",
    "normalized-losses"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_columns].dtypes)
```

```python
# 2. Horsepower vs Price
plt.figure(figsize=(12, 6))

sns.scatterplot(
    x="horsepower",
    y="price",
    color="red",
    data=df,
)
plt.title("horsepower vs price")
plt.ylabel("horsepower")
plt.xlabel("price")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3. Engine Size vs City MPG
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="engine-size",
    y="city-mpg",
    color="darkblue",
)

plt.title("Engine Size vs City MPG")
plt.xlabel("Engine Size")
plt.ylabel("City MPG")
plt.show()
```

```python
# 4. Price Distribution by Body Style
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="body-style",
    y="price",
    color="green",
)

plt.title("Price Distribution Across Body Styles")
plt.xlabel("Body Style")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```

```python
# 5. Select Technical Specifications
# ============================================

technical_specs = [
    "symboling",
    "wheel-base",
    "length",
    "width",
    "height",
    "curb-weight",
    "engine-size",
    "bore",
    "stroke",
    "compression-ratio",
    "horsepower",
    "peak-rpm",
    "city-mpg",
    "highway-mpg",
    "price"
]

technical_df = df[technical_specs]
```

```python
technical_df
```

```python
# 6. Correlation Matrix

correlation = technical_df.corr()

# 7. Correlation Heatmap


plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    cbar=False,
    square=True,
    linewidth=1
)

plt.title("Correlation Heatmap of Vehicle Technical Specifications")
plt.show()
```

```python
# 8. Correlation with Price
# ============================================

price_correlation = correlation["price"].sort_values(
    ascending=False
)

print("Correlation with Vehicle Price:")
print(price_correlation)
```

```python
# 9. Basic Market Analysis

print("\nAverage Price by Body Style:")
print(
    df.groupby("body-style")["price"]
      .mean()
      .sort_values(ascending=False)
)

print("\nAverage Horsepower by Body Style:")
print(
    df.groupby("body-style")["horsepower"]
      .mean()
      .sort_values(ascending=False)
)

print("\nAverage City MPG by Body Style:")
print(
    df.groupby("body-style")["city-mpg"]
      .mean()
      .sort_values(ascending=False)
)
```
